```python
import requests
import json
import time
import pandas as pd
from datetime import datetime, timedelta
from IPython.display import display, HTML

# === СОКРАЩЕНИЯ КОМАНД ===
TEAM_ABBR = {
    "Cleveland Guardians": "CLE", "Tampa Bay Rays": "TB",
    "Arizona Diamondbacks": "ARI", "Washington Nationals": "WSH",
    "Chicago Cubs": "CHC", "Pittsburgh Pirates": "PIT",
    "Toronto Blue Jays": "TOR", "Boston Red Sox": "BOS",
    "New York Yankees": "NYY", "Chicago White Sox": "CWS",
    "St. Louis Cardinals": "STL", "Los Angeles Angels": "LAA",
    "Houston Astros": "HOU", "Oakland Athletics": "OAK",
    "San Francisco Giants": "SFG", "Milwaukee Brewers": "MIL",
    "Philadelphia Phillies": "PHI", "Miami Marlins": "MIA",
    "Baltimore Orioles": "BAL", "Atlanta Braves": "ATL",
    "New York Mets": "NYM", "Cincinnati Reds": "CIN",
    "Seattle Mariners": "SEA", "Texas Rangers": "TEX",
    "Detroit Tigers": "DET", "Minnesota Twins": "MIN",
    "Kansas City Royals": "KC", "Colorado Rockies": "COL",
    "Los Angeles Dodgers": "LAD", "San Diego Padres": "SD",
}

# === КЭШИ ===
TEAM_ID_CACHE = {}
TEAM_ROSTER_CACHE = {}
TEAM_SCHEDULE_CACHE = {}
PLAYER_LOGS_CACHE = {}
PLAYER_STATUS_CACHE = {}

def get_team_id(team_name):
    if team_name in TEAM_ID_CACHE:
        return TEAM_ID_CACHE[team_name]
    try:
        url = f"https://statsapi.mlb.com/api/v1/teams?sportId=1"
        resp = requests.get(url)
        if resp.status_code == 200:
            data = resp.json()
            for team in data.get('teams', []):
                if team['name'] == team_name or team['teamName'] == team_name:
                    TEAM_ID_CACHE[team_name] = team['id']
                    return team['id']
    except:
        pass
    return None

def get_team_roster(team_name):
    if team_name in TEAM_ROSTER_CACHE:
        return TEAM_ROSTER_CACHE[team_name]
    team_id = get_team_id(team_name)
    if not team_id:
        return []
    try:
        url = f"https://statsapi.mlb.com/api/v1/teams/{team_id}/roster?rosterType=active&season=2026"
        resp = requests.get(url)
        if resp.status_code == 200:
            data = resp.json()
            players = []
            for player in data.get('roster', []):
                pos = player.get('position', {}).get('abbreviation', '')
                if pos not in ['P', 'RP', 'SP']:
                    players.append({
                        'id': player['person']['id'],
                        'name': player['person']['fullName']
                    })
            TEAM_ROSTER_CACHE[team_name] = players
            return players
    except:
        pass
    return []

def get_team_schedule(team_id):
    cache_key = f"{team_id}_{datetime.now().strftime('%Y-%m-%d')}"
    if cache_key in TEAM_SCHEDULE_CACHE:
        return TEAM_SCHEDULE_CACHE[cache_key]

    end_date = datetime.now().strftime("%Y-%m-%d")
    start_date = (datetime.now() - timedelta(days=60)).strftime("%Y-%m-%d")
    url = f"https://statsapi.mlb.com/api/v1/schedule?sportId=1&teamId={team_id}&startDate={start_date}&endDate={end_date}"
    try:
        resp = requests.get(url)
        if resp.status_code == 200:
            data = resp.json()
            games = []
            today = datetime.now().date()
            for date_data in data.get('dates', []):
                for game in date_data.get('games', []):
                    game_date = datetime.strptime(date_data['date'], '%Y-%m-%d').date()
                    if game_date > today:
                        continue
                    status = game.get('status', {})
                    detailed_state = status.get('detailedState', '')
                    coded_state = status.get('codedGameState', '')
                    if 'Final' in detailed_state or 'Game Over' in detailed_state or coded_state == 'F':
                        games.append({
                            'game_pk': game['gamePk'],
                            'date': date_data['date'],
                            'away': game['teams']['away']['team']['name'],
                            'home': game['teams']['home']['team']['name']
                        })
            games.sort(key=lambda x: x['date'])
            TEAM_SCHEDULE_CACHE[cache_key] = games
            return games
    except:
        pass
    return []

def get_player_game_logs(player_id):
    if player_id in PLAYER_LOGS_CACHE:
        return PLAYER_LOGS_CACHE[player_id]
    url = f"https://statsapi.mlb.com/api/v1/people/{player_id}/stats?stats=gameLog&group=hitting&season=2026"
    resp = requests.get(url)
    if resp.status_code != 200:
        return {}
    data = resp.json()
    logs = {}
    for stat_group in data.get('stats', []):
        for split in stat_group.get('splits', []):
            game_pk = split.get('game', {}).get('gamePk')
            if not game_pk:
                continue
            stat = split.get('stat', {})
            if int(stat.get('atBats', 0)) == 0:
                logs[game_pk] = None
            else:
                # Пытаемся получить синглы (ordinars)
                singles = stat.get('singles')
                if singles is None:
                    # Если поле singles отсутствует, считаем через разность
                    hits = int(stat.get('hits', 0))
                    doubles = int(stat.get('doubles', 0))
                    triples = int(stat.get('triples', 0))
                    home_runs = int(stat.get('homeRuns', 0))
                    singles = hits - doubles - triples - home_runs
                    if singles < 0:
                        singles = 0
                logs[game_pk] = {
                    'H': int(stat.get('hits', 0)),
                    '1B': singles,
                    'R': int(stat.get('runs', 0)),
                    'RBI': int(stat.get('rbi', 0)),
                    'TB': int(stat.get('totalBases', 0)),
                    'SO': int(stat.get('strikeOuts', 0)),
                    'BB': int(stat.get('baseOnBalls', 0)),
                }
    PLAYER_LOGS_CACHE[player_id] = logs
    return logs

def get_player_status(player_id):
    if player_id in PLAYER_STATUS_CACHE:
        return PLAYER_STATUS_CACHE[player_id]
    try:
        url = f"https://statsapi.mlb.com/api/v1/people/{player_id}"
        resp = requests.get(url)
        if resp.status_code == 200:
            data = resp.json()
            people = data.get('people', [])
            if people:
                person = people[0]
                status = {
                    'active': person.get('active', False),
                    'injured': person.get('injured', False),
                    'status': person.get('status', {}).get('description', 'Active')
                }
                PLAYER_STATUS_CACHE[player_id] = status
                return status
    except:
        pass
    return {'active': True, 'injured': False, 'status': 'Active'}

def analyze_player(player, team_name, match_label):
    name = player['name']
    player_id = player['id']
    print(f"  Обработка {name}...", end=' ')

    # Проверка травмы
    status = get_player_status(player_id)
    if status.get('injured', False) or 'IL' in status.get('status', '') or 'Day-to-Day' in status.get('status', ''):
        print("⚠️ травмирован")
        return None, {
            'name': name,
            'team': TEAM_ABBR.get(team_name, team_name[:3]),
            'match': match_label,
            'reason': f"травма ({status['status']})"
        }

    team_id = get_team_id(team_name)
    if not team_id:
        print("❌ нет ID команды")
        return None, None

    all_games = get_team_schedule(team_id)
    if not all_games:
        print("❌ нет завершённых матчей за 60 дней")
        return None, None

    # Матчи команды за периоды
    last_20_games = all_games[-20:] if len(all_games) >= 20 else all_games
    last_7_games = all_games[-7:] if len(all_games) >= 7 else all_games

    if len(last_20_games) < 20:
        print(f"⚠️ только {len(last_20_games)} завершённых матчей за 60 дней")

    logs = get_player_game_logs(player_id)
    if not logs:
        print("❌ нет данных")
        return None, None

    # Фильтруем сыгранные матчи игрока (где stat is not None)
    def get_played_games(games):
        played = []
        for g in games:
            game_pk = g['game_pk']
            if game_pk in logs and logs[game_pk] is not None:
                played.append(g)
        return played

    played_20_games = get_played_games(last_20_games)
    played_7_games = get_played_games(last_7_games)

    # Проверка участия в последнем матче
    if last_7_games:
        last_game_pk = last_7_games[-1]['game_pk']
        played_last = any(g['game_pk'] == last_game_pk for g in played_7_games)
        if not played_last:
            print("⚠️ не играл в последнем матче")
            return None, {
                'name': name,
                'team': TEAM_ABBR.get(team_name, team_name[:3]),
                'match': match_label,
                'reason': "не играл в последнем матче"
            }

    # Количество сыгранных матчей для отображения
    played20 = len(played_20_games)
    played7 = len(played_7_games)
    total20 = len(last_20_games)
    total7 = len(last_7_games)

    # Если игрок сыграл меньше 4 из 7 – исключаем
    if played7 < 4:
        print("⚠️ не проходит (сыграл <4 из 7)")
        return None, {
            'name': name,
            'team': TEAM_ABBR.get(team_name, team_name[:3]),
            'match': match_label,
            'reason': f"сыграл {played7} из 7"
        }

    # Функция расчёта процентов по списку сыгранных матчей (игрок точно в них играл)
    def calc_stats_from_played(played_games):
        if not played_games:
            return None
        total = len(played_games)
        stats = {
            'H>0.5': 0, 'H<0.5': 0, 'H<1.5': 0,
            'R>0.5': 0, 'R<0.5': 0,
            'RBI<0.5': 0,
            'TB<1.5': 0, 'TB<2.5': 0, 'TB<3.5': 0, 'TB>1.5': 0,
            'SO>0.5': 0, 'SO<0.5': 0, 'SO<1.5': 0, 'SO>1.5': 0,
            'BB>0.5': 0, 'BB<0.5': 0,
            '1B<0.5': 0,
        }
        for g in played_games:
            game_pk = g['game_pk']
            stat = logs[game_pk]
            if stat['H'] > 0.5:
                stats['H>0.5'] += 1
            if stat['H'] < 0.5:
                stats['H<0.5'] += 1
            if stat['H'] < 1.5:
                stats['H<1.5'] += 1
            if stat['R'] > 0.5:
                stats['R>0.5'] += 1
            if stat['R'] < 0.5:
                stats['R<0.5'] += 1
            if stat['RBI'] < 0.5:
                stats['RBI<0.5'] += 1
            if stat['TB'] < 1.5:
                stats['TB<1.5'] += 1
            if stat['TB'] < 2.5:
                stats['TB<2.5'] += 1
            if stat['TB'] < 3.5:
                stats['TB<3.5'] += 1
            if stat['TB'] > 1.5:
                stats['TB>1.5'] += 1
            if stat['SO'] > 0.5:
                stats['SO>0.5'] += 1
            if stat['SO'] < 0.5:
                stats['SO<0.5'] += 1
            if stat['SO'] < 1.5:
                stats['SO<1.5'] += 1
            if stat['SO'] > 1.5:
                stats['SO>1.5'] += 1
            if stat['BB'] > 0.5:
                stats['BB>0.5'] += 1
            if stat['BB'] < 0.5:
                stats['BB<0.5'] += 1
            if stat['1B'] < 0.5:
                stats['1B<0.5'] += 1
        return {k: (v / total * 100) for k, v in stats.items()}

    # Берём последние 7 и 20 сыгранных матчей (если их больше)
    if len(played_20_games) >= 20:
        played_20_last = played_20_games[-20:]
    else:
        played_20_last = played_20_games
    if len(played_7_games) >= 7:
        played_7_last = played_7_games[-7:]
    else:
        played_7_last = played_7_games

    stats20 = calc_stats_from_played(played_20_last)
    stats7 = calc_stats_from_played(played_7_last)

    if stats20 is None or stats7 is None:
        print("⚠️ недостаточно данных")
        return None, None

    # Формируем строки для Сыграл 20 и Сыграл 7
    played20_str = f"{played20}/{total20}"  # без значка
    played7_str = f"{played7}/{total7}"
    if played7 == 7:
        played7_str = f'<span style="border-bottom: 3px solid #00AA00;">{played7_str}</span>'
    elif played7 == 6:
        played7_str = f'<span style="border-bottom: 3px solid #FFD700;">{played7_str}</span>'
    elif played7 == 5:
        played7_str = f'<span style="border-bottom: 3px solid #FF8C00;">{played7_str}</span>'
    elif played7 == 4:
        played7_str = f'<span style="border-bottom: 3px solid #FF0000;">{played7_str}</span>'

    row = {
        'Команда': TEAM_ABBR.get(team_name, team_name[:3]),
        'Игрок': name,
        'Сыграл 20': played20_str,
        'Сыграл 7': played7_str
    }
    any_market = False
    for key in stats20.keys():
        v20 = stats20[key]
        v7 = stats7[key]
        v20_int = int(round(v20))
        v7_int = int(round(v7))
        if v20 >= 50 or v7 >= 50:
            any_market = True
            if v20 >= 70 and v7 >= 70:
                row[key] = f"✅ {v20_int}/{v7_int}%"
            elif v7 >= 70 and v20 >= 55:
                row[key] = f"🟡 {v20_int}/{v7_int}%"
            elif v20 >= 50 and v7 >= 50:
                row[key] = f"🟠 {v20_int}/{v7_int}%"
            else:
                if v20 >= 50 and v7 < 50:
                    row[key] = f"❌ {v20_int}/{v7_int}%"
                elif v7 >= 50 and v20 < 50:
                    row[key] = f"❓ {v20_int}/{v7_int}%"
    if not any_market:
        print("⚠️ нет рынков с % >=50")
        return None, None
    print("✅")
    return row, None

def format_cell_value(val):
    if not isinstance(val, str):
        return val
    if val.startswith('❌') or val.startswith('❓'):
        parts = val.split()
        if len(parts) >= 3:
            pct_str = parts[1]
            pct_parts = pct_str.replace('%', '').split('/')
            if len(pct_parts) == 2:
                try:
                    v20 = int(pct_parts[0])
                    v7 = int(pct_parts[1])
                    v20_style = 'color: red;' if v20 < 50 else ''
                    v7_style = 'color: red;' if v7 < 50 else ''
                    symbol = parts[0]
                    return f'{symbol} <span style="{v20_style}">{v20}</span>/<span style="{v7_style}">{v7}</span>%'
                except:
                    pass
    return val

def display_grouped_table(df, group_cols=['Команда']):
    if df.empty:
        print("❌ Нет данных для вывода")
        return

    df['Фамилия'] = df['Игрок'].apply(lambda x: x.split()[-1])
    df = df.sort_values(by=['Матч', 'Команда', 'Фамилия']).reset_index(drop=True)
    df.drop(columns=['Фамилия'], inplace=True)

    table_style = '''
        border-collapse: collapse;
        width: 100%;
        font-family: Arial, sans-serif;
        font-size: 13px;
        border: 1px solid #2c3e50;
    '''
    cell_style = '''
        padding: 6px 4px;
        text-align: center;
        border: 1px solid #2c3e50;
    '''

    html = []
    cols = df.columns.tolist()
    display_cols = [c for c in cols if c != 'Матч']
    html.append(f'''
    <table style="{table_style}">
        <thead>
            <tr style="background-color: #2c3e50; color: white; font-weight: bold;">
                {''.join([f'<th style="padding: 8px 4px; text-align: center; border: 1px solid #2c3e50;">{col}</th>' for col in display_cols])}
            </tr>
        </thead>
        <tbody>
    ''')

    current_group = None
    row_counter = 0

    for idx, row in df.iterrows():
        group_key = row['Команда']
        if group_key != current_group:
            match = row['Матч']
            team = row['Команда']
            html.append(f'''
                <tr style="border-top: 3px solid #FF8C00; border-bottom: 2px solid #8A2BE2; background-color: #2c3e50; color: white;">
                    <td colspan="{len(display_cols)}" style="padding: 6px 4px; text-align: center; font-weight: bold; font-size: 14px;">
                        {match} — {team}
                    </td>
                </tr>
            ''')
            html.append(f'<tr style="border-top: 3px solid #FF8C00; border-bottom: 3px solid #FF8C00; font-weight: bold; background-color: #3a4a6a;">')
            for col in display_cols:
                html.append(f'<td style="{cell_style} font-weight: bold; color: white;">{col}</td>')
            html.append('</tr>')
            current_group = group_key
            row_counter = 0
        else:
            if row_counter > 0 and row_counter % 6 == 0:
                html.append(f'<tr style="border-top: 3px solid #FF8C00; border-bottom: 3px solid #FF8C00; font-weight: bold; background-color: #3a4a6a;">')
                for col in display_cols:
                    html.append(f'<td style="{cell_style} font-weight: bold; color: white;">{col}</td>')
                html.append('</tr>')

        html.append('<tr>')
        for col in display_cols:
            val = row[col]
            formatted_val = format_cell_value(val)
            html.append(f'<td style="{cell_style}">{formatted_val}</td>')
        html.append('</tr>')
        row_counter += 1

    html.append('</tbody></table>')
    legend_html = '''
    <div style="margin-top: 10px; font-family: Arial, sans-serif; font-size: 12px; border-top: 2px solid #FF8C00; padding-top: 5px;">
        <b>Легенда:</b><br>
        <span style="border-bottom: 3px solid #00AA00;">7/7</span> – зелёное подчёркивание (все матчи)<br>
        <span style="border-bottom: 3px solid #FFD700;">6/7</span> – жёлтое подчёркивание (один пропуск)<br>
        <span style="border-bottom: 3px solid #FF8C00;">5/7</span> – оранжевое подчёркивание (два пропуска)<br>
        <span style="border-bottom: 3px solid #FF0000;">4/7</span> – красное подчёркивание (три пропуска)<br>
        <span style="margin-left: 10px;">✅ – оба % ≥70% (20/7)</span><br>
        <span style="margin-left: 10px;">🟡 – 7 ≥70%, 20 ≥55% (но <70)</span><br>
        <span style="margin-left: 10px;">🟠 – оба % ≥50% (но не зелёный и не жёлтый)</span><br>
        <span style="margin-left: 10px;">❌ – 20 ≥50%, 7 < 50%</span><br>
        <span style="margin-left: 10px;">❓ – 7 ≥50%, 20 < 50%</span><br>
        <span style="margin-left: 10px;">% указаны как: 20 / 7</span><br>
        <span style="margin-left: 10px;">Число ниже 50 выделено красным</span>
        <span style="margin-left: 10px;">Последняя колонка: 1B<0.5 – процент матчей без синглов</span>
    </div>
    '''
    html.append(legend_html)

    display(HTML(''.join(html)))

# === ОСНОВНАЯ ЧАСТЬ ===
DATE = datetime.now().strftime("%Y-%m-%d")
print(f"📅 Анализ матчей на {DATE}")

schedule_url = f"https://statsapi.mlb.com/api/v1/schedule?sportId=1&date={DATE}"
resp = requests.get(schedule_url)
if resp.status_code != 200:
    print("❌ Не удалось получить расписание")
    exit()
data = resp.json()
games = []
for d in data.get('dates', []):
    for game in d.get('games', []):
        away = game['teams']['away']['team']['name']
        home = game['teams']['home']['team']['name']
        if ('Cincinnati' in away and 'Cleveland' in home) or ('Cleveland' in away and 'Cincinnati' in home):
            continue
        game_time = game.get('gameDate', '')
        games.append({
            'game_pk': game['gamePk'],
            'away': away,
            'home': home,
            'time': game_time,
            'match_label': f"{away} @ {home}"
        })

games.sort(key=lambda x: x['time'])

all_rows = []
missed_players = []

for game in games:
    match_label = game['match_label']
    print(f"\n{'='*100}")
    print(f"Матч: {match_label} (время: {game['time']})")
    print('='*100)
    for team in [game['away'], game['home']]:
        print(f"\n  Состав команды {team}:")
        roster = get_team_roster(team)
        if not roster:
            print("    ❌ Не удалось получить состав")
            continue
        for player in roster:
            row, missed = analyze_player(player, team, match_label)
            if row:
                row['Матч'] = match_label
                all_rows.append(row)
            if missed:
                missed_players.append(missed)
            time.sleep(0.3)

if all_rows:
    df = pd.DataFrame(all_rows)
    base_cols = ['Матч', 'Команда', 'Игрок', 'Сыграл 20', 'Сыграл 7']
    market_cols = [
        'H>0.5', 'H<1.5', 'R>0.5', 'R<0.5', 'RBI<0.5',
        'SO>0.5', 'SO<1.5', 'BB<0.5',
        'TB<1.5', 'TB<2.5', 'TB<3.5', 'TB>1.5',
        'SO>1.5', 'H<0.5', 'SO<0.5', 'BB>0.5',
        '1B<0.5'
    ]
    available_markets = [c for c in market_cols if c in df.columns]
    cols_order = base_cols + available_markets
    df_display = df[cols_order]

    print("\n" + "="*100)
    print("ЛЕГЕНДА:")
    print("  🟢 - 7/7 (все матчи)")
    print("  🟡 - 6/7 (один пропуск)")
    print("  🟠 - 5/7 (два пропуска)")
    print("  🔴 - 4/7 (три пропуска)")
    print("  ✅ - зелёный: оба % ≥70% (20/7)")
    print("  🟡 - жёлтый: 7 ≥70%, 20 ≥55% (но <70)")
    print("  🟠 - оранжевый: оба % ≥50% (но не зелёный и не жёлтый)")
    print("  ❌ - красный крест: 20 ≥50%, 7 < 50%")
    print("  ❓ - красный вопрос: 7 ≥50%, 20 < 50%")
    print("  % указаны как: за 20 игр / за 7 игр")
    print("  Число ниже 50 выделено красным")
    print("  Последняя колонка: 1B<0.5 – процент матчей без синглов")
    print("="*100 + "\n")

    display_grouped_table(df_display, group_cols=['Команда'])
else:
    print("❌ Нет данных для вывода")

if missed_players:
    print("\n" + "="*100)
    print("⚠️ ИГРОКИ, НЕ ПРОШЕДШИЕ ФИЛЬТР (травма, <4 из 7 или не играли в последнем матче):")
    print("="*100)
    for p in missed_players:
        print(f"  {p['name']} ({p['team']}) — {p['match']} — причина: {p['reason']}")
```
